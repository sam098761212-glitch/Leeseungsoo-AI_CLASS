```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Sequential
from sklearn.metrics import mean_absolute_error, accuracy_score

tf.keras.utils.set_random_seed(42)

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# MNIST 딥러닝 분류: 숫자 0~9 중 하나 예측
classification_model = Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])
classification_model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
classification_model.fit(
    x_train, y_train, epochs=10, batch_size=128, validation_split=0.2,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)],
    verbose=1
)
class_pred = classification_model.predict(x_test, verbose=0).argmax(axis=1)
print("MNIST 분류 정확도:", accuracy_score(y_test, class_pred))
```

```python
regression_model = Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(1)
])
regression_model.compile(optimizer="adam", loss="mse", metrics=["mae"])
regression_model.fit(
    x_train, y_train.astype("float32"), epochs=10, batch_size=128, validation_split=0.2,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)],
    verbose=1
)
reg_pred = regression_model.predict(x_test, verbose=0).ravel()
print("MNIST 회귀 MAE:", mean_absolute_error(y_test, reg_pred))
```
